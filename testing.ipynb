```python
from src.api import DadosAbertosPBAPI
from src.extractor import BaseExtractor
from typing import Any, Dict, List

api = DadosAbertosPBAPI()
ano_target = 2025
```

```python


class AtasExtractor(BaseExtractor):
    @property
    def resource_name(self) -> str:
        return "atas"

    @property
    def period_mode(self) -> str:
        return "ano_inicio_vigencia"

    def fetch_vigencia_records(self, year: int) -> List[Dict[str, Any]]:
        return self.api.obter_todas_atas_registro_preco(ano_inicio_vigencia=year)

extrator = AtasExtractor(api=api, output_dir="./dados")
resultado = extrator.extract_and_save_year(ano_target, formats=["csv"], include_report=True)
print(f"Total: {resultado['total_registros']}")
print(resultado["arquivos"])
```
